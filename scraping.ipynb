# Scraping review aplikasi Pluang

```python
from google_play_scraper import app, reviews, Sort, reviews_all
import pandas as pd
```

Hal pertama yang dilakukan adalah melakukan import library yang akan digunakan

```python
import nltk
nltk.download('punkt_tab')
nltk.download('stopwords')
```

Melalui library nltk, kita akan mengunduh dataset yang diperlukan untuk tokenisasi teks ('punkt_tab')dan mengunduh dataset yang berisi daftar kata-kata berhenti (stop words) dalam berbagai bahasa ('stopwords').

```python
scrapreview = reviews_all(
    'com.EmasDigi',
    lang='id',
    country='id',
    sort=Sort.MOST_RELEVANT,
    count=3500
)
```

Selanjutnya adalah melakukan scraping melalui ID 'com.EmasDigi'. Pada kode diatas, 'id' dimasukkan sebagai bahasa dan negara. Untuk urutan ulasan digunakan Sort.MOST_RELEVANT dengan jumlah maksimum ulasan yang diambil bernilai 3500

```python
import csv

with open('dataUlasan.csv', mode='w', newline='',encoding='utf-8') as file:
    writer = csv.writer(file)
    writer.writerow(['Review'])
    for review in scrapreview:
        writer.writerow([review['content']])
```

Terakhir adalah menyimpan ulasan/hasil scraping pada file CSV dengan nama dataUlasan.csv
